```python
from pathlib import Path
import jax
import jax.numpy as jnp
import matplotlib.pyplot as plt
from psilia import get_console
from psilia.vision import (
    CameraIntrinsics,
    project, 
    unproject
)


console = get_console()

key = jax.random.PRNGKey(0)
intr = CameraIntrinsics.load(Path("~/workspace/data/temp/intr0.yaml").expanduser())

console.print(intr)
```

```python
key,_ = jax.random.split(key)
N = 10
uvs = jax.random.uniform(key, (N, 2), minval=jnp.zeros(2), maxval=jnp.array([intr.w, intr.h]))
zs = jax.random.uniform(key, (N,1), minval=0.1, maxval=10.0)

xs = unproject(jnp.concatenate([uvs, zs], axis=-1), intr)

uvs_, _ = project(xs, intr)


# =============================================
fig, axs = plt.subplots(1, 2, figsize=(10, 5))
axs[0].set_aspect("equal")
axs[0].set_xlim(0, intr.w)
axs[0].set_ylim(0, intr.h)
axs[0].scatter(uvs[:, 0], uvs[:, 1], c="r", s=10)
axs[1].set_aspect("equal")
axs[1].set_xlim(0, intr.w)
axs[1].set_ylim(0, intr.h)
axs[1].scatter(uvs_[:, 0], uvs_[:, 1], c="r", s=10)
```

```python
factor = 0.5

intr_ = CameraIntrinsics(
    intr.intrinsics * factor,
    intr.distortion_coeffs,
    intr.camera_model,
    intr.distortion_model,
    int(intr.width * factor),
    int(intr.height * factor),
    intr.camera_name
)

uvs_, _ = project(xs, intr_)


# =============================================
fig, axs = plt.subplots(1, 2, figsize=(10, 5))
axs[0].set_aspect("equal")
axs[0].set_xlim(0, intr.w)
axs[0].set_ylim(0, intr.h)
axs[0].scatter(uvs[:, 0], uvs[:, 1], c="r", s=10)
axs[1].set_aspect("equal")
axs[1].set_xlim(0, intr.w*factor)
axs[1].set_ylim(0, intr.h*factor)
axs[1].scatter(uvs_[:, 0], uvs_[:, 1], c="r", s=10)
```
