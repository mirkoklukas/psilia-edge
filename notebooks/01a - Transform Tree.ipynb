```python
from psilia.transforms.transform import Transform
from psilia.transforms.transform_tree import Tree, TransformTree
from psilia import get_console
import jax
import jax.numpy as jnp


console = get_console()
key = jax.random.PRNGKey(0)
```

```python
# TODO: Make this nice
def sample_transform(key, t0=jnp.zeros(3), q0=jnp.array([0.0,0.0,0.0,1.0])):
    t = t0 + jax.random.normal(key,(3,) )
    q = q0 + jax.random.normal(key,(4,) )
    q = q/jnp.linalg.norm(q, axis=-1)
    return Transform(t,q)

key,*keys = jax.random.split(key,4)
tf0 = sample_transform(keys[0])
tf1 = sample_transform(keys[1])
tf2 = sample_transform(keys[2])

tf0.parent="a"
tf0.child ="b"

tf1.parent="b"
tf1.child ="c"

tf2.parent="b"
tf2.child ="d"
```

```python
console.print(TransformTree([tf0, tf1, tf2]))
```
